Problem 1: Mobile Recharge Validation System


```python
def validate_recharge(amount):
    valid_plans = [199, 299, 399, 599]

    if amount < 50:
        print("❌ Recharge amount must be at least ₹50")
        return False

    if amount not in valid_plans:
        print("❌ Invalid plan. Available plans: 199, 299, 399, 599")
        return False

    print("✅ Recharge successful!")
    return True


while True:
    recharge_amount = int(input("Enter recharge amount: "))

    if validate_recharge(recharge_amount):
        break
    else:
        print("Please try again.\n")
```

Problem 2: Inventory Reorder Alert System


```python
def check_inventory(inventory):
  for products,stock in inventory.items():
    if stock<15:
      print(products ,"restock alert")
    else:
      print(products,"sufficient stock exists")
inventory={
   "rice":10,
    "milk":25,
   "wheat":20,
   "chocolates":5
}
check_inventory(inventory)
```

Problem 4: Cab Fare Estimator with Retry Option

```python
def calculate_fare(distance, travel_time):

    base_fare = 50
    per_km_rate = 12

    # peak hours list
    peak_hours = ["8am", "9am", "10am", "5pm", "6pm", "7pm"]


    fare = base_fare + (distance * per_km_rate)

    # check if travel time is peak hour
    if travel_time in peak_hours:
        print("Peak hour charges applied (25% extra)")
        fare += 0.25 * fare

    return fare



while True:

    distance = float(input("Enter distance (km): "))
    time = input("Enter travel time (e.g., 8am, 3pm): ").lower()

    total_fare = calculate_fare(distance, time)
    print("Total Fare: ₹", total_fare)
    retry = input("Do you want to calculate again? (yes/no): ").lower()

    if retry != "yes":
        break
```

Problem 5: Employee Attendance Eligibility Checker


```python
def attendance_eligibility(attendance):
  count=0
  for i in attendance:

    if i=="P":
      count+=1;
  attendance_percentage=(count/len(attendance))*100
  if attendance_percentage>=75:
      return "Eligible"
  else:
      return "Not Eligible"
attendance=["P","A","P","A","P"]
result=attendance_eligibility(attendance)
print(result)
```

Problem 6: Password Strength Checker


```python
def password_strength(password):
  password_length=len(password)
  if password_length<8:
    return "password length must be at least 8"
  has_digit=False
  has_special=False
  for ch in password:

        if ch.isdigit():   # check digit
            has_digit = True

        if ch in "@#$":    # check special character
            has_special = True
  if has_digit and has_special:
        return "Strong Password"
  else:
        return "Weak Password (Must contain digit and @#$)"
password=input("enter the password")
print(password_strength(password))
```
